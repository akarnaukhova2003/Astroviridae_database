```python
import pandas as pd

csv_file = "../constants/Astroviridae_Aves_Amphibia_14072026_host.csv"
ids_file = "../../clade_analysis/AVES/final_files/Aves_full_ids_removed.txt"

output_file = "../constants/Astroviridae_Aves_Amphibia_14072026_host_filtered.csv"


def normalize_id(x):
    if pd.isna(x):
        return None

    return str(x).strip().upper().replace("-", "_")


df = pd.read_csv(csv_file)

with open(ids_file) as f:
    ids = {
        normalize_id(line)
        for line in f
        if line.strip()
    }

df["ID_normalized"] = df["ID"].apply(normalize_id)

filtered_df = df[df["ID_normalized"].isin(ids)].copy()

filtered_df = filtered_df.drop(columns=["ID_normalized"])

filtered_df.to_csv(output_file, index=False)

# Проверка
csv_ids = set(df["ID_normalized"].dropna())
matched_ids = csv_ids & ids
missing_ids = ids - csv_ids

print(f"Всего строк в CSV: {len(df)}")
print(f"ID в TXT: {len(ids)}")
print(f"Совпадений: {len(matched_ids)}")
print(f"Строк после фильтрации: {len(filtered_df)}")
print(f"ID из TXT, отсутствующие в CSV: {len(missing_ids)}")
print(f"Результат сохранён: {output_file}")
```

```python
from Bio import Phylo
from pathlib import Path
import re

input_file = Path("../../Aves_Amphibia_trees/Aves_2_97_aa_2306_after_trim_iqtree_clr.nex")
output_file = Path("../../Aves_Amphibia_trees/Aves_2_97_aa_2306_after_trim_iqtree_clr_pastml.nwk")

tree = Phylo.read(input_file, "nexus")

for clade in tree.find_clades():
    if clade.comment:
        match = re.search(r"label=(\d+)", clade.comment)
        if match:
            clade.confidence = float(match.group(1))
        clade.comment = None

Phylo.write(tree, output_file, "newick", plain=False)

check_tree = Phylo.read(output_file, "newick")

print("Количество терминальных узлов:", len(check_tree.get_terminals()))
print("Количество bootstrap-значений:", sum(
    clade.confidence is not None for clade in check_tree.get_nonterminals()
))
print("Первый узел:", check_tree.get_terminals()[0].name)
print("Файл:", output_file)
```

```python
from Bio import Phylo
from pathlib import Path

input_file = Path("../../Aves_Amphibia_trees/Aves_2_97_aa_2306_after_trim_iqtree_clr_pastml.nwk")
output_file = Path("../../Aves_Amphibia_trees/Aves_2_97_aa_2306_after_trim_iqtree_clr_pastml_fixed.nwk")

tree = Phylo.read(input_file, "newick")

# Удаляем лишние кавычки в названиях терминальных узлов
for tip in tree.get_terminals():
    if tip.name:
        tip.name = tip.name.rstrip("'").lstrip("'")

Phylo.write(tree, output_file, "newick", plain=False)

# Проверка
check_tree = Phylo.read(output_file, "newick")

print("Терминальных узлов:", len(check_tree.get_terminals()))
print("Bootstrap:", sum(c.confidence is not None for c in check_tree.get_nonterminals()))

print("\nПервые 5 ID:")
for tip in check_tree.get_terminals()[:5]:
    print(repr(tip.name))

print("\nФайл:", output_file)
```

```python
import pandas as pd
from Bio import Phylo

tree_file = "../../Aves_Amphibia_trees/Aves_2_97_aa_2306_after_trim_iqtree_clr_pastml_fixed.nwk"
csv_file = "../constants/Astroviridae_Aves_Amphibia_14072026_host_filtered_trimmed.csv"

tree = Phylo.read(tree_file, "newick")
df = pd.read_csv(csv_file, sep="\t", dtype=str)

tree_ids = {tip.name.split("/")[0] for tip in tree.get_terminals()}
csv_ids = set(df["ID"].dropna().str.strip())

missing_ids = sorted(tree_ids - csv_ids)

print("Совпадений:", len(tree_ids & csv_ids))
print("Нет в CSV:", len(missing_ids))
print("Лишние ID в CSV:", len(csv_ids - tree_ids))

print("\nID из дерева, которых нет в CSV:")
for id_ in missing_ids:
    print(id_)

with open("missing_ids.txt", "w") as f:
    f.write("\n".join(missing_ids))

print("\nСписок сохранён в missing_ids.txt")
```
